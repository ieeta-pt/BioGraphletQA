```python
import json
```

```python
with open("prompt_builder.json") as f:
    shards = json.load(f)
```

```python
shards.keys()
```

```python
prompts = [
    ['baseline'], #baseline
    ['baseline', 'simple_example'], #baseline with ex
    #testing now qa instructions
    ['baseline','instructions_markdown', 'question_instruction', 'answer_instruction'], #baseline qa instructions
    ['baseline','instructions_markdown', 'question_instruction', 'answer_instruction', 'simple_example'], # + example
    # testing now analysis instrcutions:
    ['baseline', 'instructions_markdown', 'analyze_graphlet_instruction', 'final_analysis'],
    #node templates
    ['baseline', 'instructions_markdown', 'analyze_graphlet_instruction', 'node_types', 'final_analysis'],
    ['baseline', 'instructions_markdown', 'analyze_graphlet_instruction', 'node_types', 'simple_example','final_analysis'], # example
    ['baseline', 'instructions_markdown', 'analyze_graphlet_instruction', 'node_types', 'question_instruction', 'answer_instruction', 'final_analysis'], #QA insturctiosn
    ['baseline', 'instructions_markdown', 'analyze_graphlet_instruction', 'node_types', 'question_instruction', 'answer_instruction', 'simple_example','final_analysis'], #QA insturctiosn + answer
    # refelction
    ['baseline','instructions_markdown','reflection_instruction'],
    ['baseline','instructions_markdown','reflection_instruction', 'complex_example'], # + example
    ['baseline','instructions_markdown','reflection_instruction', 'question_evaluation', 'answer_evaluation'], # + qa evaluation
    ['baseline','instructions_markdown','reflection_instruction', 'question_evaluation', 'answer_evaluation', 'complex_example'], # + qa evaluation + example
    ['baseline','instructions_markdown','question_instruction', 'answer_instruction','reflection_instruction', 'question_evaluation', 'answer_evaluation', 'complex_example','final_analysis'], # + qa evaluation + example # +qa instructoins
    #all
    ['baseline','instructions_markdown','analyze_graphlet_instruction', 'node_types','question_instruction', 'answer_instruction','reflection_instruction', 'question_evaluation', 'answer_evaluation', 'complex_example', 'final_analysis'], 

   ]
```

```python
i = 1

final_prompts = {}
for i in range(len(prompts)):
    prompt=""
    instruction_counter = 1 
    for part in prompts[i]:
        if 'instruction' in part and part!= 'instructions_markdown':
            prompt+=f"{instruction_counter}. "
            instruction_counter+=1 
        prompt+=shards[part]+"\n"
    prompt+=shards['json']+"\n"
    final_prompts[i] = prompt
# print(prompt)
```

```python
with open(f'prompt_templates.json', 'w') as fp:
            json.dump(final_prompts, fp)
```
